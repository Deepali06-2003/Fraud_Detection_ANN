## Data cleaning

```python
first_chunk["timestamp"] = pd.to_datetime(
    first_chunk["timestamp"], errors="coerce"
)

first_chunk["transaction_dayofweek"] = (
    first_chunk["timestamp"].dt.dayofweek
)

first_chunk["transaction_month"] = (
    first_chunk["timestamp"].dt.month
)
```

```python
first_chunk = first_chunk.dropna(
    subset=["timestamp", "is_fraud"]
)

first_chunk = first_chunk.sort_values("timestamp").reset_index(drop=True)

n = len(first_chunk)
train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_first_chunk = first_chunk.iloc[:train_end].copy()
val_first_chunk = first_chunk.iloc[train_end:val_end].copy()
test_first_chunk = first_chunk.iloc[val_end:].copy()

print("Train:", train_first_chunk.shape)
print("Validation:", val_first_chunk.shape)
print("Test:", test_first_chunk.shape)
```

```python
target = "is_fraud"

drop_cols = [
    "is_fraud",
    "transaction_id",
    "customer_id",
    "card_number",
    "timestamp",
    "merchant",
    "device_fingerprint",
    "ip_address"
]

X_train = train_first_chunk.drop(
    columns=drop_cols, errors="ignore"
)
y_train = train_first_chunk[target].astype("int8")

X_val = val_first_chunk.drop(
    columns=drop_cols, errors="ignore"
)
y_val = val_first_chunk[target].astype("int8")

X_test = test_first_chunk.drop(
    columns=drop_cols, errors="ignore"
)
y_test = test_first_chunk[target].astype("int8")
```

```python
categorical_cols = X_train.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

numerical_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
```

## preprocessing and train test split

```python
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from joblib import parallel_backend

# Reduce memory usage
X_train[numerical_cols] = X_train[numerical_cols].astype("float32")
X_val[numerical_cols] = X_val[numerical_cols].astype("float32")

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=100,
        sparse_output=True,
        dtype=np.float32
    ))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numerical_cols),
        ("categorical", categorical_pipeline, categorical_cols)
    ],
    sparse_threshold=1.0,
    n_jobs=-1
)

# Fit only on training data
with parallel_backend("loky"):
    X_train_processed = preprocessor.fit_transform(X_train)

# Transform validation data without refitting
X_val_processed = preprocessor.transform(X_val)

print("Train shape:", X_train_processed.shape)
print("Validation shape:", X_val_processed.shape)
print("Sparse output:", hasattr(X_train_processed, "tocsr"))
```

```python
#calculating class weights
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = {
    0: weights[0],
    1: weights[1]
}

print(class_weights)
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Dense, BatchNormalization, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

def build_ann(input_dim, optimizer="adam", hidden_layers=2):

    model = Sequential()
    model.add(Input(shape=(input_dim,)))

    if hidden_layers == 1:
        units = [64]
    elif hidden_layers == 2:
        units = [64, 32]
    elif hidden_layers == 3:
        units = [128, 64, 32]
    else:
        raise ValueError("hidden_layers must be 1, 2, or 3")

    for i, neurons in enumerate(units):
        model.add(Dense(neurons, use_bias=False))
        model.add(BatchNormalization())
        model.add(tf.keras.layers.Activation("sigmoid"))

        dropout_rate = 0.3 if i == 0 else 0.2
        model.add(Dropout(dropout_rate))

    model.add(Dense(1, activation="sigmoid"))

    if optimizer.lower() == "adam":
        opt = tf.keras.optimizers.Adam(learning_rate=0.001)
    elif optimizer.lower() == "rmsprop":
        opt = tf.keras.optimizers.RMSprop(learning_rate=0.001)
    else:
        raise ValueError("optimizer must be 'adam' or 'rmsprop'")

    model.compile(
        optimizer=opt,
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
            tf.keras.metrics.AUC(name="roc_auc"),
            tf.keras.metrics.AUC(name="pr_auc", curve="PR")
        ]
    )

    return model
```

## final model

```python
import tensorflow as tf
import numpy as np

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Dense, BatchNormalization,
    Activation, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

# Reproducibility
tf.keras.utils.set_random_seed(42)

# Three hidden layers: 128, 64, 32 neurons
model = Sequential([
    Input(shape=(X_train_processed.shape[1],)),

    Dense(128, use_bias=False),
    BatchNormalization(),
    Activation("sigmoid"),
    Dropout(0.3),

    Dense(64, use_bias=False),
    BatchNormalization(),
    Activation("sigmoid"),
    Dropout(0.2),

    Dense(32, use_bias=False),
    BatchNormalization(),
    Activation("sigmoid"),
    Dropout(0.2),

    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="roc_auc"),
        tf.keras.metrics.AUC(name="pr_auc", curve="PR")
    ]
)

model.summary()

early_stop = EarlyStopping(
    monitor="val_pr_auc",
    mode="max",
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    X_train_processed,
    y_train,
    validation_data=(X_val_processed, y_val),
    epochs=20,
    batch_size=512,
    class_weight=class_weights,
    callbacks=[early_stop],
    verbose=1
)
```

```python
from sklearn.metrics import (
    precision_recall_curve,
    f1_score
)

y_val_prob = model.predict(
    X_val_processed,
    batch_size=4096
).ravel()

precision, recall, thresholds = precision_recall_curve(
    y_val, y_val_prob
)

# The final precision/recall entries have no corresponding threshold
f1_scores = (
    2 * precision[:-1] * recall[:-1] /
    (precision[:-1] + recall[:-1] + 1e-12)
)

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

print("Best threshold:", best_threshold)
print("Validation precision:", precision[best_idx])
print("Validation recall:", recall[best_idx])
print("Validation F1:", f1_scores[best_idx])
```

## Streaming Pipeline for Full Dataset Training
To train on the entire dataset without exhausting RAM or GPU memory, we will:
1. Create a Python generator that reads the CSV in chunks (`chunksize=100000`).
2. Apply the same feature engineering (timestamp conversion, day of week, month extraction) to each chunk.
3. Drop unneeded columns and use the already fitted `preprocessor` to transform the chunk into a sparse/dense feature matrix.
4. Convert the data to a streaming `tf.data.Dataset` using `tf.data.Dataset.from_generator` to supply batches continuously to the model.

```python
import pandas as pd
import numpy as np
import tensorflow as tf

def dataset_generator(csv_path, chunk_size=100000, is_training=True):
    """Reads the CSV chunk-by-chunk, preprocesses it, and yields features and labels."""
    while True:
        # Read chunk by chunk
        for chunk in pd.read_csv(csv_path, chunksize=chunk_size):
            # 1. Feature Engineering
            chunk["timestamp"] = pd.to_datetime(chunk["timestamp"], errors="coerce")
            chunk["transaction_dayofweek"] = chunk["timestamp"].dt.dayofweek.fillna(-1)
            chunk["transaction_month"] = chunk["timestamp"].dt.month.fillna(-1)

            # Clean null values in target if training
            chunk = chunk.dropna(subset=["is_fraud"])

            # Separate features and target
            y = chunk["is_fraud"].astype("int8").values
            X = chunk.drop(columns=drop_cols, errors="ignore")

            # Match float types
            X[numerical_cols] = X[numerical_cols].astype("float32")

            # 2. Apply preprocessor (already fitted earlier)
            X_processed = preprocessor.transform(X)

            # Convert sparse matrix from preprocessor to dense array for Keras
            if hasattr(X_processed, "toarray"):
                X_processed = X_processed.toarray()

            # Yield each row or mini-batches
            for i in range(len(y)):
                yield X_processed[i], y[i]

        # If not training, we only need to go through the dataset once
        if not is_training:
            break
```

```python
# Determine the output signature based on our processed shapes
output_signature = (
    tf.TensorSpec(shape=(X_train_processed.shape[1],), dtype=tf.float32),
    tf.TensorSpec(shape=(), dtype=tf.int8)
)

# Create the tf.data.Dataset wrapper around the generator
full_train_dataset = tf.data.Dataset.from_generator(
    generator=lambda: dataset_generator(DATASET_PATH, chunk_size=100000, is_training=True),
    output_signature=output_signature
)

# Batch and prefetch the data to speed up GPU computation
BATCH_SIZE = 1024
full_train_dataset = full_train_dataset.shuffle(buffer_size=10000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

print("Streaming training dataset initialized successfully!")
```

### Train Neural Network on Full Streamed Dataset
Now, we will train the neural network. Since we are streaming the dataset using an infinite generator, we must explicitly define `steps_per_epoch`. Given our dataset size is roughly 7.48M rows, and we are using a batch size of 1024, one epoch requires roughly:
`7,483,766 / 1024 ≈ 7300 steps`.

```python
# Reset model parameters or rebuild the model architecture
import os
import gc
tf.keras.backend.clear_session()
gc.collect()

full_model = build_ann(
    input_dim=X_train_processed.shape[1],
    optimizer="adam",
    hidden_layers=3
)

# Train for a few epochs across the entire dataset without running out of RAM
# We will use class_weights computed earlier to address the 80/20 class imbalance
steps_per_epoch = 5000

history_full = full_model.fit(
    full_train_dataset,
    epochs=3, # Stream through entire 7.48M dataset 3 times
    steps_per_epoch=steps_per_epoch,
    class_weight=class_weights,
    verbose=1
)

# Ensure the output directory exists before saving
os.makedirs("fraud_detection_model", exist_ok=True)

# Save the full trained model
full_model.save("fraud_detection_model/full_dataset_fraud_model.keras")
print("Finished training and saved the full dataset model!")
```

## Final Evaluation of the Full-Dataset Model
Now that we have successfully saved the final neural network trained on the entire 7.48-million-row streaming dataset, let's load it, optimize its decision threshold, and evaluate its performance on the test partition.

```python
import tensorflow as tf
import json
import numpy as np
from sklearn.metrics import (
    precision_recall_curve,
    classification_report,
    confusion_matrix,
    average_precision_score,
    roc_auc_score,
    f1_score
)

# 1. Load the model trained on the full dataset
full_dataset_model = tf.keras.models.load_model("fraud_detection_model/full_dataset_fraud_model.keras")

# 2. Predict on validation data to determine the best threshold
y_val_prob_full = full_dataset_model.predict(X_val_processed, batch_size=4096).ravel()
precision_full, recall_full, thresholds_full = precision_recall_curve(y_val, y_val_prob_full)

# Compute F1 scores to find the threshold that balances precision and recall
f1_scores_full = (2 * precision_full[:-1] * recall_full[:-1]) / (precision_full[:-1] + recall_full[:-1] + 1e-12)
best_idx_full = np.argmax(f1_scores_full)
best_threshold_full = thresholds_full[best_idx_full]

print(f"Best Decision Threshold: {best_threshold_full:.4f}")
print(f"Validation Precision: {precision_full[best_idx_full]:.4f}")
print(f"Validation Recall: {recall_full[best_idx_full]:.4f}")
print(f"Validation F1 Score: {f1_scores_full[best_idx_full]:.4f}\n")

# Save this newly optimized threshold alongside our final model
with open("fraud_detection_model/full_dataset_threshold.json", "w") as f:
    json.dump({"threshold": float(best_threshold_full)}, f, indent=4)
```

```python
# 3. Predict on the unseen test set to evaluate true performance
y_test_prob_full = full_dataset_model.predict(X_test_processed, batch_size=4096).ravel()
y_test_pred_full = (y_test_prob_full >= best_threshold_full).astype(int)

# 4. Generate metrics
print("=== Classification Report (Full Model) ===")
print(classification_report(y_test, y_test_pred_full, target_names=["Legitimate", "Fraud"]))

print("=== Confusion Matrix ===")
print(confusion_matrix(y_test, y_test_pred_full))

print(f"\nTest PR-AUC: {average_precision_score(y_test, y_test_prob_full):.4f}")
print(f"Test ROC-AUC: {roc_auc_score(y_test, y_test_prob_full):.4f}")
print(f"Test F1-Score: {f1_score(y_test, y_test_pred_full):.4f}")
```
